11. Fill missing values in Sleep_Hours, Study_Hours, Social_Media_Hours, and Attendance using median imputation. Why median over mean here?

```python
data['Sleep_Hours'] = data['Sleep_Hours'].fillna(data['Sleep_Hours'].median())
data['Study_Hours'] = data['Study_Hours'].fillna(data['Study_Hours'].median())
data['Social_Media_Hours'] = data['Social_Media_Hours'].fillna(data['Social_Media_Hours'].median())
data['Attendance'] = data['Attendance'].fillna(data['Attendance'].median())

```

In students lifestyle dataset filling the missing values with median values because some student may study for less hours and some students may study for very long hours which will impact the average of the study hours. so choosing median is better option.

14. Detect outliers in Sleep_Hours, Study_Hours, and Social_Media_Hours using the IQR method. How many outliers exist in each?

```python
# function to check and handle outliers

def handle_outliers_with_median(df, column_name):
    # 1. Calculate IQR statistics using clean, local variable names
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    IQR = Q3 - Q1
    
    # 2. Define bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # 3. Find and count outliers
    outliers_mask = (df[column_name] < lower_bound) | (df[column_name] > upper_bound)
    outlier_count = outliers_mask.sum()
    print(f"Found {outlier_count} outliers in '{column_name}'")
    
    # # 4. Replace outliers with the column median
    # median_val = df[column_name].median()
    # df.loc[outliers_mask, column_name] = median_val
    
    return df
```

```python
data = handle_outliers_with_median(data,'Sleep_Hours')
data = handle_outliers_with_median(data,'Study_Hours')
data = handle_outliers_with_median(data,'Social_Media_Hours')
```
